Requirement:

a. Predict student Result using Decision Tree classifier based on student data attributes

b. Visualize the data as a decision tree

c. Compute the accuracy score of the mode

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn import tree
import matplotlib.pyplot as plt
```

a. Predict student Result using Decision Tree classifier based on student data attributes

```python
students = pd.read_csv('student_data.csv')

# Sort by Result in a defined order before analysis
grade_order = ['A', 'B', 'C', 'Fail']
students['Result'] = pd.Categorical(students['Result'], categories=grade_order, ordered=True)
students = students.sort_values('Result').reset_index(drop=True)

X = students[['Attendance', 'Assignment','MidExam','StudyHours','GPA','Participation']]
y = students['Result']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=75)
clf = DecisionTreeClassifier(random_state=75)
clf.fit(X_train, y_train)
```

```python
y_pred=clf.predict(X_test)
```

```python
y_test
```

b. Visualize the data as a decision tree

```python
plt.figure(figsize=(14,10))
tree.plot_tree(clf, feature_names=X.columns, class_names=grade_order, filled=True, label='all', fontsize=8)
plt.show()
```

c. Compute the accuracy score of the mode

```python
accuracy=accuracy_score(y_test,y_pred)
print("Accuracy:", accuracy*100, "%")
```
