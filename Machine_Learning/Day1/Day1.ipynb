```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn import tree
import matplotlib.pyplot as plt

#loading the dataset
iris = load_iris()
X=iris.data
y=iris.target

#converting the dataset into a pandas dataframe
df=pd.DataFrame(np.c_[iris['data'],iris['target']],columns=iris['feature_names']+['target'])
df
```

```python
print(y)
```

```python
#splitting the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=21)
```

```python
#Training the Decision Tree Classifier

#Initializing the Decision Tree Classifier
clf=DecisionTreeClassifier(random_state=21)

#Training the model
clf.fit(X_train,y_train)
```

```python
#predicting the target values for the test set
y_pred=clf.predict(X_test)

#calculate the accuracy of the model
accuracy=accuracy_score(y_test,y_pred)
print(f"Accuracy: {accuracy:.2f}")
```

```python
#visualizing the decision tree
#plot the tree

plt.figure(figsize=(14,8))
tree.plot_tree(clf, feature_names=iris['feature_names'], class_names=iris['target_names'],label='all',filled=True,fontsize=10)
plt.show()
```
