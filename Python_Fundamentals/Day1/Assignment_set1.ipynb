8. Shallow vs Deep Copy

• Create a nested list

• Make a shallow copy and modify an inner element

• Make a deep copy and modify an inner element

• Print all lists and explain the difference

```python
import copy

# Must use a NESTED list to see the difference
original = [[1, 2], [3, 4], [5, 6]]

# Shallow copy — outer list is new, but inner lists are shared
shallow = copy.copy(original)
shallow[0][0] = 99          # modifies the shared inner list
print(f"original: {original}")   # ← also changed!
print(f"shallow:  {shallow}")

print()

# Deep copy — fully independent, nothing is shared
original2 = [[1, 2], [3, 4], [5, 6]]
deep = copy.deepcopy(original2)
deep[0][0] = 99             # only affects deep copy
print(f"original2: {original2}")  # ← unchanged
print(f"deep:      {deep}")
```
