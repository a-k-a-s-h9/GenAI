## 1.1 Variables and Dynamic Typing

Python uses **dynamic typing** — you don't declare types. The interpreter infers them.

Key built-in types:
- `int`, `float`, `complex` — numbers
- `str` — text
- `bool` — True/False
- `NoneType` — absence of value

```python
# ── Dynamic Typing ──────────────────────────────────────────────────────────
name    = "Alice"         # str
age     = 30              # int
salary  = 85000.50        # float
is_active = True          # bool
bonus   = None            # NoneType
print(f"{name} | age={age} | salary={salary} | active={is_active} | bonus={bonus}")
print(f"Types: {type(name)} | {type(age)} | {type(salary)} | {type(is_active)} | {type(bonus)}")
```

```python
# Variable reassignment changes type — Python allows this
x = 10
print(f"x = {x} → type: {type(x)}")
x = "10"
print(f"x = {x} → type: {type(x)}")
```

```python
# ── Multiple Assignment (Pythonic style) ─────────────────────────────────────
a, b, c = 10, 20, 30
print(f"a={a}, b={b}, c={c}")
```

```python
x="Sachin"
print(x)
print("Player name is " + x)
print("Player name is",x)
print("Player name is %s"%x)
print("Player name is {}".format(x))
print(f"Player name is {x}")
```

## 1.2 Operators

| Category | Operators |
|---|---|
| Arithmetic | `+  -  *  /  //  %  **` |
| Comparison | `==  !=  >  <  >=  <=` |
| Logical | `and  or  not` |
| Identity | `is , is not` |
| Membership | `in , not in` |
| Bitwise | `&  |  ^  ~  <<  >>` |

```python
# ── Arithmetic ────────────────────────────────────────────────────────────────
print(17 // 5)   # floor division → 3
print(17 %  5)   # modulo         → 2
print(2  ** 8)   # exponentiation → 256

# ── Comparison & Logical ─────────────────────────────────────────────────────
score = 78
print(score >= 50 and score < 90)   # True
print(not (score > 90))             # True
```

```python
# ── Identity vs Equality ─────────────────────────────────────────────────────
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a == b)   # True  — same values
print(a != b)   # False — same values
print(a is b)   # False — different objects
print(a is  c)   # True  — same object
print(a is not b)   # True  — different objects
```

## 1.3 Strings — The Most Used Type in Data/AI Work

Strings are **immutable sequences** of Unicode characters.  
Key operations: slicing, formatting, methods, multiline.

```python
# ── String Creation ───────────────────────────────────────────────────────────
s1 = 'single quotes'
s2 = "double quotes"  #  "double"
s3 = """Multi-line
string example"""
raw = r"C:\Users\Documents\newfile.txt"   # raw string — backslash not treated as escape
print(raw)
```

```python
# ── Indexing & Slicing ────────────────────────────────────────────────────────
text = "Python Programming"
print(text[0])       # P
print(text[-1])      # g
print(text[0:6])     # Python
print(text[7:])      # Programming
print(text[::-1])    # Reverse: gnimmargorP nohtyP
```

```python
# ── Key String Methods ────────────────────────────────────────────────────────
s = "  Hello, World!  "
print(s.strip())            # remove whitespace
print(s.strip().lower())    # lowercase
print(s.strip().replace(",", "").replace("!", ""))  # clean text
print("python".upper().center(20, '-'))  # -----PYTHON-----
```

```python
# ── f-strings (modern formatting — preferred) ────────────────────────────────
name, score = "Sachin", 95.7
print(f"{name} scored {score:.1f}%")          # Sachin scored 95.7%
print(f"{name!r} → repr")                     # 'Sachin' → repr
print(f"{'='*40}")
print(f"{'Centered Title':^40}")
```

```python
# ── REAL-WORLD USE CASE: Log Parsing ─────────────────────────────────────────
# Scenario: Parse server access logs to extract IP, method, endpoint, status

log_entry = '192.168.1.10 - - [09/Jun/2026:10:23:45 +0000] "GET /api/users HTTP/1.1" 200 1024'
parts = log_entry.split('"')
print(parts)
ip        = log_entry.split()[0]
request   = parts[1] # "GET /api/users HTTP/1.1"
method=request.split()[0]  # GET
endpoint  = request.split()[1]      # /api/users
status    = parts[2].strip().split()[0]     # 200

print(f"IP: {ip}")
print(f"Method: {method}")
print(f"Endpoint: {endpoint}")
print(f"Status: {status}")
print(f"Is success: {status.startswith('2')}")
```

## 2.1 Lists — Ordered, Mutable Sequences

Most versatile collection. Supports heterogeneous elements, indexing, slicing, iteration.

```python
# ── List Basics ───────────────────────────────────────────────────────────────
fruits = ["apple", "banana", "cherry", "date"]

# CRUD on list
fruits.append("mango")          # add to end
fruits.insert(1, "avocado")          # insert at index
fruits.remove("banana")              # remove by value
popped = fruits.pop()                # remove & return last
print(fruits)
print(f"Popped: {popped}")

# Sorting
numbers = [5, 1, 9, 3, 7, 2]
numbers.sort()                        # in-place ascending
print(numbers)
print(sorted(numbers, reverse=True)) # sorted() → new list
print(numbers)
# Nested list (matrix)
matrix = [[1, 2, 3],
           [4, 5, 6],
           [7, 8, 9]]
print(f"Row 1: {matrix[0]}")
print(f"Center element: {matrix[1][1]}")
```

```python
# ── REAL-WORLD USE CASE: Employee Roster Management ──────────────────────────
employees = [
    {"id": 101, "name": "Alice",   "dept": "Engineering", "salary": 90000},
    {"id": 102, "name": "Bob",     "dept": "Marketing",   "salary": 72000},
    {"id": 103, "name": "Charlie", "dept": "Engineering", "salary": 95000},
    {"id": 104, "name": "Diana",   "dept": "HR",          "salary": 68000},
    {"id": 105, "name": "Eve",     "dept": "Engineering", "salary": 88000},
]
# Sort by salary descending
top_earners = sorted(employees, key=lambda e: e["salary"], reverse=True)
print("Top earners:")
for emp in top_earners[:3]:
    print(f"  {emp['name']:10} — ₹{emp['salary']:,}")

# Filter Engineering dept
eng_team = [e for e in employees if e["dept"] == "Engineering"]
print(eng_team)
avg_eng_salary = sum(e["salary"] for e in eng_team) / len(eng_team)
print(f"\nEngineering avg salary: ₹{avg_eng_salary:,.0f}")
```

```python
# Named tuples — self-documenting records
from collections import namedtuple

Employee = namedtuple('Employee', ['name', 'department', 'salary'])
emp = Employee(name='Alice', department='Engineering', salary=90000)
print(f"\nNamed Tuple: {emp}")
print(f"Name: {emp.name}, Dept: {emp.department}")
```

```python
# Convert between list and tuple
my_list  = [1, 2, 3]
my_tuple = tuple(my_list)
print(f"\nList→Tuple: {my_tuple}")

# Convert back from tuple to list
new_list = list(my_tuple)
print(f"Tuple→List: {new_list}")
```

## 2.3 Sets — Unordered, Unique Elements

Sets are backed by a hash table — O(1) membership tests.  
Use for: **deduplication, intersection, union, difference**.

```python
# ── Set Operations ────────────────────────────────────────────────────────────
skills_alice = {"Python", "SQL", "Machine Learning", "Docker"}
skills_bob   = {"Python", "JavaScript", "Docker", "React"}

print(f"Union (all skills):         {skills_alice | skills_bob}")
print(f"Intersection (common):      {skills_alice & skills_bob}")
print(f"Difference (Alice only):    {skills_alice - skills_bob}")
print(f"Sym. Diff (not in both):    {skills_alice ^ skills_bob}")
```

```python
# Deduplication use case
raw_data = [1, 2, 2, 3, 3, 3, 4, 5, 5]
raw_data_set = set(raw_data)
unique   = list(raw_data_set)
print(f"\nOriginal: {raw_data}")
print(f"Unique:   {sorted(unique)}")
```

```python

# Deduplication use case
raw_data = [1, 2, 2, 3, 3, 3, 4, 5, 5]
unique   = list(set(raw_data))
print(f"\nOriginal: {raw_data}")
print(f"Unique:   {sorted(unique)}")
```

```python
# Membership test — O(1) for sets vs O(n) for lists
banned_users = {"spammer1", "hacker99", "bot_user"}
user = "hacker99"
print(f"\n{user} is banned: {user in banned_users}")
```

## 2.4 Dictionaries — Key-Value Store

Python's most powerful built-in data structure. Ordered (Python 3.7+), mutable, O(1) access.

```python
# ── Dictionary Basics ─────────────────────────────────────────────────────────
config = {
    "host":     "localhost",
    "port":     5432,
    "database": "prod_db",
    "ssl":      True,
}
# Access
print(config["host"])                     # direct access — KeyError if missing
print(config.get("timeout", 30))          # safe access with default

# Mutate
config["max_connections"] = 100           # add
config["port"] = 5433                     # update
config.pop("ssl")                         # remove
```

```python
config = {
    "host":     "localhost",
    "port":     5432,
    "database": "prod_db",
    "ssl":      True,
}
# Iterate
for element in config.items():
    print(element)
print('-' * 40)

for key in config:
    print(key)

print('-' * 40)
for key in config.keys():
    print(key)

print('-' * 40)

for value in config.values():
    print(value)
```

```python

for key,value in config.items():
    print(key,value)
```

```python
# Merge dicts (Python 3.9+)
defaults   = {"timeout": 30, "retries": 3}
overrides  = {"timeout": 60}
final_conf = defaults | overrides       # override takes precedence
print(f"\nMerged: {final_conf}")
```

```python
# dict.setdefault — avoid KeyError when building nested dict
groups = {}
for item in ["apple", "avocado", "banana", "blueberry", "cherry"]:
    groups.setdefault(item[0], []).append(item)
print(f"\nGrouped: {groups}")
```

```python
# ── REAL-WORLD USE CASE: Word Frequency Counter (NLP pre-processing) ──────────
from collections import Counter, defaultdict

text = """
machine learning is transforming industries. machine learning models require data.
data science and machine learning are closely related fields. 
deep learning is a subset of machine learning.
"""
# Tokenise & count
words      = text.lower().split()
print(f"Words: {words}")
stop_words = {"is", "a", "of", "and", "are"}
clean      = [w.strip('.,') for w in words if w.strip('.,') not in stop_words]
print(f"Cleaned Words: {clean}")
freq = Counter(clean)
print(f"\nWord Frequencies: {freq}")
print("Top 5 frequent words:")
for word, count in freq.most_common(5):
    bar = '█' * count
    print(f"  {word:20} {bar} ({count})")
```

## 3.1 List Comprehensions

**Syntax:** `[expression for item in iterable if condition]`

More readable and faster than equivalent `for` loops for simple transformations.

```python
# ── Basic Comprehensions ──────────────────────────────────────────────────────
squares  = [x**2 for x in range(1, 11)]
print(f"Squares: {squares}")

evens    = [x for x in range(21) if x % 2 == 0]
print(f"Evens: {evens}")

upper    = [s.upper() for s in ["hello", "world", "python"]]
print(f"Uppercase: {upper}")

# With if-else (ternary inside expression)
labels = ["PASS" if score >= 50 else "FAIL" for score in [45, 78, 32, 90, 55]]
print(f"\nLabels:   {labels}")

# Nested comprehension — flatten a 2D matrix
matrix   = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat     = [cell for row in matrix for cell in row]  
#  Read it like this: For each row in matrix, for each cell in that row, collect cell
print(f"Flattened: {flat}")
```

```python
names  = ["alice", "bob", "charlie"]
scores = [85, 92, 78]
zipped=zip(names, scores)
print(list(zipped))
```

```python
# Dict comprehension
names  = ["alice", "bob", "charlie"]
scores = [85, 92, 78]
grade_map = {name: score for name, score in zip(names, scores)}
print(grade_map)
```

```python
# Set comprehension — unique domains from email list
emails  = ["alice@gmail.com", "bob@yahoo.com", "charlie@gmail.com", "dave@hotmail.com"]
domains = {email.split('@')[1] for email in emails}
print(f"Unique domains: {domains}")
```

```python
# ── REAL-WORLD USE CASE: Data Cleaning Pipeline ───────────────────────────────
# Scenario: Clean a raw CSV-like dataset of product records

raw_products = [
    {"id": 1,  "name": "  Laptop  ",  "price": "₹85000", "category": "electronics"},
    {"id": 2,  "name": "Chair",        "price": "₹4500",  "category": "FURNITURE"},
    {"id": 3,  "name": "",             "price": "₹200",   "category": "accessories"},
    {"id": 4,  "name": "Headphones",   "price": "₹3200",  "category": "Electronics"},
    {"id": 5,  "name": "Desk",         "price": "UNKNOWN","category": "furniture"},
]
print(raw_products)
# Clean in one comprehension pipeline
cleaned = [
    {
        "id":       p["id"],
        "name":     p["name"].strip().title(),
        "price":    int(p["price"].replace("₹", "")) if p["price"].replace("₹","").isdigit() else None,
        "category": p["category"].lower(),
    }
    for p in raw_products
    if p["name"].strip()          # exclude records with empty name
]

print(f"{'ID':<5} {'Name':<15} {'Price':>10} {'Category'}")  # :<5  Print text left-aligned in a width of 5 characters
print('-' * 45)
for p in cleaned:
    price_str = f"₹{p['price']:,}" if p['price'] else "N/A"
    print(f"{p['id']:<5} {p['name']:<15} {price_str:>10} {p['category']}")
```

## 3.2 Tuple Unpacking & Starred Expressions

Unpacking lets you **destructure iterables** into named variables — cleaner than indexing.

```python
# ── Basic Unpacking ───────────────────────────────────────────────────────────
point = (10.5, 20.3)
x, y  = point                     # tuple unpacking
print(f"x={x}, y={y}")
```

```python
a,b=10,20
print(f"a={a}, b={b}")
a,b= b,a
print(f"After swap: a={a}, b={b}")
```

```python
# Unpack nested
((lat, lon), altitude) = ((28.6139, 77.2090), 216)
print(f"Delhi — lat:{lat}, lon:{lon}, alt:{altitude}m")
```

```python
a,*b=[1,2,3,4,5]
print(a)
print(b)
```

```python
# ── Starred Expressions (*) ──────────────────────────────────────────────────
first, *rest = [1, 2, 3, 4, 5,6,7,8,9]
print(f"\nfirst={first}, rest={rest}")
```

```python
*head, last = [10, 20, 30, 40]
print(f"head={head}, last={last}")
```

```python
first, *middle, last = range(1, 8)
print(f"first={first}, middle={list(middle)}, last={last}")
```

```python
a, *b, c = range(1, 8)
print(f"a={a}, b={b}, c={c}")
```

```python
# ── Unpacking in Loops ────────────────────────────────────────────────────────
transactions = [
    ("TXN001", "2026-06-01", 1500.00, "credit"),
    ("TXN002", "2026-06-02",  320.50, "debit"),
    ("TXN003", "2026-06-03", 8750.00, "credit"),
]
print("\nTransaction Summary:")
total_credit = 0
for txn_id, date, amount, txn_type in transactions:   # unpack in loop!
    print(f"  [{txn_id}] {date}  ₹{amount:>10,.2f}  ({txn_type})")
    if txn_type == "credit":
        total_credit += amount
print(f"  Total credits: ₹{total_credit:,.2f}")
```

```python
# ── REAL-WORLD USE CASE: Unpacking API Response ───────────────────────────────
# Scenario: Parse structured API response from a weather service

def get_weather_data(city: str):
    """Simulates an API response."""
    return {
        "city": city,
        "coords": (28.6, 77.2),
        "conditions": ("Partly Cloudy", 34, 65),   # (desc, temp_C, humidity%)
        "forecast": [(36, 28), (35, 27), (33, 25)], # (max, min) per day
    }

data = get_weather_data("Delhi")
city              = data["city"]
lat, lon          = data["coords"]                   # unpack coords tuple
desc, temp, humid = data["conditions"]               # unpack conditions
today, *rest_days = data["forecast"]                 # starred — today + remaining
today_max, today_min = today 

print(f"City       : {city} ({lat}°N, {lon}°E)")
print(f"Conditions : {desc}, {temp}°C, Humidity: {humid}%")
print(f"Today      : Max {today_max}°C / Min {today_min}°C")
print(f"Next days  : {rest_days}")
```
