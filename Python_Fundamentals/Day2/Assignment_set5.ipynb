Import pandas as pd.

```python
import pandas as pd
```

Read Salaries.csv as a dataframe called sal

```python
df = pd.read_csv('Salaries.csv')
```

Check the head of the DataFrame.

```python
df.head()
```

Use the .info() method to find out how many entries there are.

```python
df.info()
```

What is the average BasePay ?

```python
base_pay = df['BasePay'].mean()
print(base_pay)
```

What is the highest amount of OvertimePay in the dataset ?

```python
highest_paid = df['OvertimePay'].max()
print(highest_paid)
```

What is the job title of JOSEPH DRISCOLL ? Note: Use all caps, otherwise you may get
an answer that doesn't match up (there is also a lowercase Joseph Driscoll).

```python
print(df[df['EmployeeName'] == 'JOSEPH DRISCOLL']['JobTitle'])
```

How much does JOSEPH DRISCOLL make (including benefits)?

```python
print(df[df['EmployeeName'] == 'JOSEPH DRISCOLL']['TotalPayBenefits'])
```

What is the name of highest paid person (including benefits)?

```python
print(df[df['TotalPayBenefits'] == df['TotalPayBenefits'].max()])
```

What is the name of lowest paid person (including benefits)? Do you notice something strange about how much he or she is paid?

```python
print(df[df['TotalPayBenefits'] == df['TotalPayBenefits'].min()])
```

What was the average (mean) BasePay of all employees per year? (2011-2014) ?

```python
print(df[(df['Year'] >= 2011) & (df['Year'] <= 2014)].groupby('Year')['BasePay'].mean())
```

How many unique job titles are there?

```python
print(df['JobTitle'].nunique())
```

What are the top 5 most common jobs?

```python
print(df.groupby('JobTitle')['JobTitle'].count().sort_values(ascending=False).head(5))
```

How many Job Titles were represented by only one person in 2011? (e.g. Job Titles
with only one occurence in 2011?)

```python
print(((df[df['Year'] == 2011]['JobTitle'].value_counts())==1).sum())
```
