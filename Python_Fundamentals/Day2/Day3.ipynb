## Addressing in Python

```python
a=[[1,2],[3,4]]
b=a

print(id(a),id(b))          #same (outer list)
print(id(a[0]),id(b[0]))    #same (inner lists)
a[0][0]=100
print(b)
```

## Shallow Copy

```python
import copy

a=[[1,2],[3,4]]
b=copy.copy(a)              # Shallow copy
print(id(a),id(b))          #different (outer list)
print(id(a[0]),id(b[0]))    #same (inner lists)
a[0][0]=100
print(b)
```

```python
import copy

original = [
    {"name": "Alice", "skills": ["Python", "Data Analysis"]},
    {"name": "Bob", "skills": ["Java", "Web Development"]}
]

shallow = copy.copy(original)  # Shallow copy
original[0]["skills"].append("Machine Learning")
print("Shallow Copy:", shallow)  # Affected by changes to mutable objects
print("Original:", original)
```

## Deep Copy

```python
import copy
a=[[1,2],[3,4]]
b=copy.deepcopy(a)          # Deep copy
print(id(a),id(b))          #different (outer list)
print(id(a[0]),id(b[0]))    #different (inner lists)
a[0][0]=100
print(b)
```

```python
import copy

original = [
    {"name": "Alice", "skills": ["Python", "Data Analysis"]},
    {"name": "Bob", "skills": ["Java", "Web Development"]}
]
deep = copy.deepcopy(original)  # Deep copy
original[0]["skills"].append("Machine Learning")
print("Deep Copy:", deep)  # Not affected by changes to mutable objects
print("Original:", original)
```

## Decorators

```python
def our_home():
    print("This is our home")

def hello_decorator(func):
    def inner():
        print("*****+++++*****")
        func()
        print("*****+++++*****")
    return inner

our_home=hello_decorator(our_home)  #our_home-> inner

our_home()
```

```python
def hello_decorator(func):
    def inner():
        print("*****+++++*****")
        func()
        print("*****+++++*****")
    return inner

@hello_decorator
def our_home():
    print("This is our home")

our_home()

@hello_decorator
def relative_home():
    print("This is our relative home")


relative_home()
```

```python
def authentication_required(func):
    def wrapper():
        # Check if user is authenticated
        user_authenticated =  False  # Simulating an unauthenticated user 
        if user_authenticated:            
            return func()
        else:            
            return "Access denied. Please log in to continue."
    return wrapper

@authentication_required
def my_secure_function():        
    return "Secret information only for authenticated users."

print(my_secure_function())
```

```python
def do_twice(func):
    def wrapper():
        func()
        func()
    return wrapper

@do_twice
def greet():
    print("Hello!")
    
greet()

#Example of a decorator that takes an argument
def greet_name(name):
    print(f"Hello, {name}!")

greet_name("Alice")
```

## Args(*) and kwargs(**)

```python
def do_twice(func):
    def wrapper_do_twice(*n,**m):   #Accepting any number of positional and keyword arguments
        func(*n,**m)
        func(*n,**m)
    return wrapper_do_twice

@do_twice
def say_hello():
    print("Hello")

say_hello()

@do_twice
def display(n):
    print("Hello", n)

display("Alice")

@do_twice
def kw_display(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
print('-------------    ')
kw_display(name="Alice", age=30, city="New York")
```
